```python
#Import necessary libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf
import keras as k
import sklearn.datasets
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, BatchNormalization, Dropout
from tensorflow.keras.optimizers import SGD, Adam
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
```

```python
# pima=sklearn.datasets.load_diabetes()
# x_data=pd.DataFrame(pima.data,columns=pima.feature_names)
# y_data=pd.Series(pima.target)
# x_train, x_test, y_train, y_test = train_test_split(x_data, y_data, test_size=0.2, random_state=42)
# x_train, x_val, y_train, y_val = train_test_split(x_train, y_train, test_size=0.2, random_state=42)
url = "https://raw.githubusercontent.com/jbrownlee/Datasets/master/pima-indians-diabetes.data.csv"
column_names = ["Pregnancies", "Glucose", "BloodPressure", "SkinThickness",
                "Insulin", "BMI", "DiabetesPedigreeFunction", "Age", "Outcome"]
pima_df = pd.read_csv(url, header=None, names=column_names)
```

```python
print(pima_df.head())
```

```python
x_data = pima_df.drop("Outcome", axis=1)
y_data = pima_df["Outcome"]
x_train, x_test, y_train, y_test = train_test_split(x_data, y_data, test_size=0.2, random_state=42)
x_train, x_val, y_train, y_val = train_test_split(x_train, y_train, test_size=0.2, random_state=42)
x_train.shape, y_train.shape
```

```python
x_test.shape, y_test.shape
```

```python
x_val.shape, y_val.shape
```

```python
scaler = StandardScaler()
x_train = scaler.fit_transform(x_train)
x_val = scaler.transform(x_val)
x_test = scaler.transform(x_test)
```

```python
model1Hid = Sequential()
model1Hid.add(Dense(128, activation='relu', input_shape=(x_train.shape[1],)))
model1Hid.add(BatchNormalization())
model1Hid.add(Dropout(0.3))
model1Hid.add(Dense(1, activation='sigmoid'))
```

```python
model1Hid.summary()
```

```python
model1Hid.compile(optimizer='SGD', loss='binary_crossentropy', metrics=['accuracy'])
model1Hid.fit(x_train, y_train, epochs=10, batch_size=32, validation_data=(x_val, y_val))
```

```python

loss, accuracy = model1Hid.evaluate(x_test, y_test)
print("Test loss: ", loss)
print("Test accuracy: ",accuracy)
```

```python
model = Sequential([
    Dense(128, activation='relu', input_shape=(x_train.shape[1],)),
    BatchNormalization(),
    Dropout(0.3),
    Dense(64, activation='relu'),
    BatchNormalization(),
    Dropout(0.3),
    Dense(1, activation='sigmoid')
])
```

```python
model.summary()
```

```python
sgd_optimizer = SGD(learning_rate=0.01)
adam_optimizer = Adam(learning_rate=0.001)
model.compile(optimizer=adam_optimizer, loss='binary_crossentropy', metrics=['accuracy'])
model.fit(x_train, y_train, epochs=10, batch_size=32, validation_data=(x_val, y_val))
```

```python
loss, accuracy = model.evaluate(x_test, y_test)
print("Test Loss:", loss)
print("Test Accuracy:", accuracy)
```

```python

from sklearn.datasets import load_iris
iri = load_iris()
x_data = pd.DataFrame(iri.data, columns=iri.feature_names)
y_data = pd.Series(iri.target)
x_data.head(), y_data.head()
```

```python
x_train, x_test, y_train, y_test = train_test_split(x_data, y_data, test_size=0.2, random_state=42)
```

```python
multi1Model = Sequential([
    Dense(128, activation='relu', input_shape=(x_train.shape[1],)),
    BatchNormalization(),
    Dropout(0.3),
    Dense(3, activation='softmax')
])
```

```python
multi1Model.summary()
```

```python
multi1Model.compile(optimizer='adam',
                    loss='sparse_categorical_crossentropy',
                    metrics=['accuracy'])

# Train the model
history = multi1Model.fit(x_train, y_train,
                          epochs=10,
                          batch_size=16,
                          validation_split=0.2)
```

```python
# Evaluate the model
loss, accuracy = multi1Model.evaluate(x_test, y_test)
print("Test Loss:", loss)
print("Test Accuracy:",accuracy)
```
