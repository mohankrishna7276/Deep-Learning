```python
import numpy as np

def gradient_descent(X, y, learning_rate, num_iterations):
    num_samples, num_features = X.shape
    theta = np.zeros(num_features)  # Initialize the theta parameters to zeros

    for _ in range(num_iterations):
        # Calculate the hypothesis/predictions
        h = np.dot(X, theta)

        # Calculate the error/residuals
        error = h - y

        # Update theta using gradient descent
        gradient = np.dot(X.T, error) / num_samples
        theta -= learning_rate * gradient

    return theta

# Example usage
X = np.array([[1, 3], [1, 4], [1, 5], [1, 6]])  # Features matrix
y = np.array([2, 4, 5, 6])  # Target vector
learning_rate = 0.1
num_iterations = 1000

# Call the gradient descent function
theta = gradient_descent(X, y, learning_rate, num_iterations)

print("Theta parameters:", theta)
```

```python
from sklearn.datasets import load_breast_cancer
cancer=load_breast_cancer()
df=pd.DataFrame(cancer.data,columns=cancer.feature_names)
print(df.isnull().sum().sum())
x=df
y=pd.Series(cancer.target)
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.3,shuffle=True)
```

```python
from sklearn.linear_model import LogisticRegression
log_reg=LogisticRegression(max_iter=2500)
log_reg.fit(x_train,y_train)
```

```python
from sklearn.metrics import accuracy_score
y_pred=log_reg.predict(x_test)
acc=accuracy_score(y_pred,y_test)
print(acc)
```

```python
from keras.models import Sequential
from tensorflow.keras.layers import InputLayer
from tensorflow.keras.layers import Dense
np.random.seed(42)
tf.random.set_seed(42)
```

```python
ANN_model=Sequential()
ANN_model.add(InputLayer(input_shape=(30,)))
#ANN_model.add(Dense(10,activation="sigmoid"))
ANN_model.add(Dense(1,activation="sigmoid"))
```

```python
ANN_model.summary()
```

```python
optimizer=tf.keras.optimizers.Adam(learning_rate=0.05)
# Compiling model with optimizer, loss, metrics.
ANN_model.compile(optimizer = "Adam", loss = "categorical_crossentropy", metrics = [k.metrics.Accuracy, k.metrics.Precision, k.metrics.F1Score])
```

```python
history=ANN_model.fit(x_train,y_train,epochs=10,batch_size=32,validation_split=0.2,shuffle=False)
```

```python
test_loss,test_acc=ANN_model.evaluate(x_test,y_test)
print(test_loss)
print(test_acc)
```
